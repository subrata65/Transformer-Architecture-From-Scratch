POSITIONAL ENCODING :-WE TRY TO ASSINGE EVERY UNIQ POSITION WITH UNIQ VALUE WHICH WILL GENARATE BY POSITIONAL ENCODING,WITH THE HELP OF THIS METHOD, THIS MODEL DO NOT MISS ARRANGE THE WORD POSITION 

```python
import torch.nn as nn
import torch
import math

d_model = 384
seq_len = 60  

pe = torch.zeros(seq_len, d_model)

for pos in range(seq_len):         
    for i in range(d_model):        
        if i % 2 == 0:
            pe[pos, i] = math.sin(pos / (10000 ** (i / d_model)))
        else:
            pe[pos, i] = math.cos(pos / (10000 ** ((i - 1) / d_model)))

print("Positional Encoding:", pe)
print("Shape:", pe.shape)   
```

1)D_MODEL MEANS THAT DIMENTION OF EVERY TOKEN
2)SEQ_LEN IS HOW MANY TOKENS WE WANT TO FIND ATTENTION
3)BATCH_SIZE MEANS THAT HOW MANY NUMBER OF ROW OF DATASET WE TRY TO SEND IN TRAING

SELF ATTENTION:-AFTER CREATING EMBEDDING  WE GENARATE THREE RANDON MATRIX WITH SAME DIMENTION AND NOW FOR EACH TOKEN WE MULTIPLAI WITH THIS THREE VECTOR AND NOW FIRST MATRIX WE WILL CALL Q(QUARY) VECTOR AND 2ND VECTOR WE CALL K(KEY) AND 3RD VECTOR WE WILL CALL V(VALUE) VECTOR 
NOW  QUARY VECTOR GOES TO EVERY KEY VECTOR AND DOT PRODUCT AND WE SCALE WITH DIMENTION OD KEY VECTOR  THEN SOFTMAX ACTIVATION FUNTION THEN WE MATRIX MULTIPLICATION WITH  VALUE VECTOR 

WHAT ACTUALLY HAPPING THERE:-WHEN WE DOT PRODUCT WITH Q AND K WE TRY TO FIND RELATION BETWEEN TWO MATRIX  

SCALING FACTOR:-Var(y)=a^2Var(x) in this formula we can see that if dimension of matrix increase then the variance between two matrix also increase for this reason we scale 

THEN WE APPLY SOFTMAX TO FIND OUT THE PROBABILTY AND THEN MULTIPAI WITH V VECTOR 

refreance:-https://arxiv.org/pdf/1706.03762

```python
import torch
import torch.nn as nn
import math

torch.manual_seed(1200)

d_model = 384
num_heads = 12
seq_len = 60
batch_size = 8

"""embeddings = torch.as_tensor(embeddings, dtype=torch.float32)   

final_token = embeddings + pe                                   
print("Final Shape (before batch):", final_token.shape)

# Add batch dimension -> (8, 60, 384)
final_token = final_token.unsqueeze(0).repeat(batch_size, 1, 1)
print("Final Shape (with batch):", final_token.shape)
print(embeddings.shape)"""
embeddings = torch.as_tensor(embeddings, dtype=torch.float32)

embeddings = embeddings[:seq_len]          # (60, 384)
final_token = embeddings + pe[:seq_len]    # (60, 384) + (60, 384)
print("Final Shape (before batch):", final_token.shape)

final_token = final_token.unsqueeze(0).repeat(batch_size, 1, 1)   # (8, 60, 384)
print("Final Shape (with batch):", final_token.shape)

#Q,K,V
class CreateMatrix(nn.Module):
    def __init__(self, d_model):
        super().__init__()
        # scaled init avoids huge attention scores (randn without scaling saturates softmax)
        self.q = nn.Parameter(torch.randn(d_model, d_model) / math.sqrt(d_model))
        self.k = nn.Parameter(torch.randn(d_model, d_model) / math.sqrt(d_model))
        self.v = nn.Parameter(torch.randn(d_model, d_model) / math.sqrt(d_model))
        self.d_model = d_model

    def forward(self, x):
        Q = torch.matmul(x, self.q)
        K = torch.matmul(x, self.k)
        V = torch.matmul(x, self.v)
        return Q, K, V

class SelfAttention(nn.Module):
    def forward(self, Q, K, V):
        dk = K.shape[-1]#SCALLING FACCTOR
        scores = torch.matmul(Q, K.transpose(-2, -1)) / math.sqrt(dk)
        attention_weights = torch.softmax(scores, dim=-1)
        output = torch.matmul(attention_weights, V)
        return output, attention_weights


class MultiHeadAttention(nn.Module):
    def __init__(self, d_model, num_heads):
        super().__init__()
        assert d_model % num_heads == 0
        self.d_model = d_model
        self.num_heads = num_heads
        self.head_dim = d_model // num_heads

        self.create_matrix = CreateMatrix(d_model)
        self.attention = SelfAttention()
        self.W_o = nn.Linear(d_model, d_model)

    def forward(self, x):
        batch_size, seq_len, d_model = x.shape

        Q, K, V = self.create_matrix(x)

        # (B, T, D) -> (B, H, T, head_dim)
        Q = Q.view(batch_size, seq_len, self.num_heads, self.head_dim).transpose(1, 2)
        K = K.view(batch_size, seq_len, self.num_heads, self.head_dim).transpose(1, 2)
        V = V.view(batch_size, seq_len, self.num_heads, self.head_dim).transpose(1, 2)

        output, attention_weights = self.attention(Q, K, V)

        # (B, H, T, head_dim) -> (B, T, D)
        output = output.transpose(1, 2).contiguous().view(batch_size, seq_len, self.d_model)

        output = self.W_o(output)
        return output, attention_weights



mha = MultiHeadAttention(d_model, num_heads)

final_token = final_token[:, :seq_len, :].to(device)
mha = mha.to(device)

output, attention_weights = mha(final_token)

print("Output shape:", output)                      
print("Attention weights shape:", attention_weights.shape) 
print(seq_len)
```

NOW THIS IS VERY CRUSIAL POINT
WHAT IS MASK ATTENTION AND WHY WE USE IT:-
MASK MATRIX:-UPPER TRAIANGUALR MATRIX BECOME "-inf"
WE TRY TO STOP FUTURE TOKEN WITH MASK MATRIX FOR STOP DATA LEAKGING 
MODEL CAN PREDICT THE TOKEN BY ITSELF

```python
import math
import torch
import torch.nn as nn

torch.manual_seed(1200)

d_model = 384
num_heads = 12
seq_len = 60
batch_size = 8


class QKV(nn.Module):
    def __init__(self, d_model):
        super().__init__()
        self.q_m = nn.Linear(d_model, d_model)
        self.k_m = nn.Linear(d_model, d_model)
        self.v_m = nn.Linear(d_model, d_model)

    def forward(self, x):
        return self.q_m(x), self.k_m(x), self.v_m(x)


class MaskedAttention(nn.Module):
    def forward(self, Q, K, V, mask=None):
        dk = K.shape[-1]
        scores = torch.matmul(Q, K.transpose(-2, -1)) / math.sqrt(dk)

        if mask is not None:
            scores = scores.masked_fill(mask == 0, float('-inf'))

        attention_weights = torch.softmax(scores, dim=-1)
        mask_output = torch.matmul(attention_weights, V)
        return mask_output, attention_weights


class MultiHeadAttention(nn.Module):
    def __init__(self, d_model, num_heads):
        super().__init__()
        assert d_model % num_heads == 0

        self.d_model = d_model
        self.num_heads = num_heads
        self.head_dim = d_model // num_heads

        self.create_matrix = QKV(d_model)         
        self.attention = MaskedAttention()        
        self.W_o = nn.Linear(d_model, d_model)

    def forward(self, x, mask=None):            
        B, T, _ = x.shape

        Q, K, V = self.create_matrix(x)


        Q = Q.view(B, T, self.num_heads, self.head_dim).transpose(1, 2)
        K = K.view(B, T, self.num_heads, self.head_dim).transpose(1, 2)
        V = V.view(B, T, self.num_heads, self.head_dim).transpose(1, 2)

        output, attention_weights = self.attention(Q, K, V, mask)

        # (B, H, T, head_dim) -> (B, T, D)
        output = output.transpose(1, 2).contiguous().view(B, T, self.d_model)

        output = self.W_o(output)
        return output, attention_weights

final_token = torch.randn(batch_size, seq_len, d_model)   
mask = torch.tril(torch.ones(seq_len, seq_len))          
mask = mask.unsqueeze(0).unsqueeze(0)                   

mha = MultiHeadAttention(d_model, num_heads)
output, attention_weights = mha(final_token, mask)

print("Output shape:", output.shape)                       
print("Attention weights shape:", attention_weights.shape) 
print(output)
```

VERY IMPORTANT PART:
DECODER OUTPUT VECTOR COMES IN PICTURE
WHAT IT ACTUALLY DO :
DECODER OUTPUT VECTOR BECOMES NOW QUARY VECTOR AND ALL THE SELF ATTENTION AND MULTI HEAD ATTENTION HAPPEN AGING WITH  KEY AND VALUE VECTOR OF ENCODER
EVERY ENGLISH WORD QUARY GOES TO BENGALI'S KEY  VECTOR  AND THEN ALL SCALLING ,SOFTMAX AND VALUE VECTOR MULTIPLICATION HAPPEN THERE NOW MODEL CAN TALK ENGLISH TO BENGALI


```python
class AddNorm(nn.Module):
    def __init__(self, d_model):
        super().__init__()
        self.layer_norm = nn.LayerNorm(d_model)

    def forward(self, X, sublayer_output):
        tt=self.layer_norm(X + sublayer_output)
        return tt


class CrossAttention(nn.Module):
    def __init__(self, d_model, num_heads):
        super().__init__()
        assert d_model % num_heads == 0
        self.d_model = d_model
        self.num_heads = num_heads
        self.head_dim = d_model // num_heads

        self.W_q = nn.Linear(d_model, d_model, bias=False)
        self.W_k = nn.Linear(d_model, d_model, bias=False)
        self.W_v = nn.Linear(d_model, d_model, bias=False)

        self.attention = MaskedAttention()   
        self.W_o = nn.Linear(d_model, d_model)

    def forward(self, decoder_x, encoder_output, src_mask=None):
        B, tgt_len, _ = decoder_x.shape
        src_len = encoder_output.shape[1]

        Q = self.W_q(decoder_x).view(B, tgt_len, self.num_heads, self.head_dim).transpose(1, 2)
        K = self.W_k(encoder_output).view(B, src_len, self.num_heads, self.head_dim).transpose(1, 2)
        V = self.W_v(encoder_output).view(B, src_len, self.num_heads, self.head_dim).transpose(1, 2)

        out, attn_weights = self.attention(Q, K, V, src_mask)

        out = out.transpose(1, 2).contiguous().view(B, tgt_len, self.d_model)
        return self.W_o(out), attn_weights
```

```python
class AddNorm(nn.Module):
    def __init__(self,out,tt):
        super().__init__()
        self.layer_norm = nn.LayerNorm(out)

    def forward(self, tt,out):
        rr=tt+out
        return self.layer_norm(rr)
```

```python

class FeedForward(nn.Module):
    def __init__(self, rr, d_ff, dropout=0.3):
        super().__init__()
        self.linear1 = nn.Linear(rr, d_ff)
        self.act = nn.GELU()
        self.dropout = nn.Dropout(dropout)
        self.linear2 = nn.Linear(d_ff, rr)

    def forward(self, x):
        x = self.linear1(x)
        x = self.act(x)
        x = self.dropout(x)
        x = self.linear2(x)
        return x
```

DECODER BLOCK

```python
class DecoderBlock(nn.Module):
    def __init__(self, d_model, num_heads, d_ff, dropout=0.1):
        super().__init__()
        self.self_attn = MultiHeadAttention(d_model, num_heads)
        self.add_norm1 = AddNorm(d_model, dropout)

        self.cross_attn = CrossAttention(d_model, num_heads)
        self.add_norm2 = AddNorm(d_model, dropout)

        self.ffn = FeedForward(d_model, d_ff, dropout)
        self.add_norm3 = AddNorm(d_model, dropout)

    def forward(self, x, encoder_output, tgt_mask=None, src_mask=None):
      
        attn_out, _ = self.self_attn(x, tgt_mask)
        x = self.add_norm1(x, attn_out)

       
        cross_out, cross_weights = self.cross_attn(x, encoder_output, src_mask)
        x = self.add_norm2(x, cross_out)
        
        ffn_out = self.ffn(x)
        x = self.add_norm3(x, ffn_out)

        return x, cross_weights
```

```python
import math, torch, torch.nn as nn
from torch.utils.data import Dataset, DataLoader, random_split
from torch.nn.utils.rnn import pad_sequence
import sentencepiece as spm
PAD_ID, BOS_ID, EOS_ID = 0, 2, 3
VOCAB_SIZE = sp.get_piece_size()
MAX_LEN = 64
DEVICE = "cuda" if torch.cuda.is_available() else "cpu"

class TranslationDataset(Dataset):
    def __init__(self, en_texts, bn_texts):
        self.en, self.bn = list(en_texts), list(bn_texts)

    def __len__(self):
        return len(self.en)

    def __getitem__(self, i):
        src = sp.encode(str(self.en[i]))[:MAX_LEN - 1] + [EOS_ID]
        tgt = sp.encode(str(self.bn[i]))[:MAX_LEN - 1]
        tgt_in  = [BOS_ID] + tgt
        tgt_out = tgt + [EOS_ID]
        return (torch.tensor(src), torch.tensor(tgt_in), torch.tensor(tgt_out))

def collate(batch):
    src, tgt_in, tgt_out = zip(*batch)
    return (pad_sequence(src,     batch_first=True, padding_value=PAD_ID),
            pad_sequence(tgt_in,  batch_first=True, padding_value=PAD_ID),
            pad_sequence(tgt_out, batch_first=True, padding_value=PAD_ID))

full = TranslationDataset(df["en"], df["bn"])
n_val = int(0.05 * len(full))
train_ds, val_ds = random_split(full, [len(full) - n_val, n_val],
                                generator=torch.Generator().manual_seed(1200))
train_loader = DataLoader(train_ds, batch_size=64, shuffle=True,  collate_fn=collate)
val_loader   = DataLoader(val_ds,   batch_size=64, shuffle=False, collate_fn=collate)
```

```python
class MultiHeadAttention(nn.Module):
    def __init__(self, d_model, num_heads):
        super().__init__()
        assert d_model % num_heads == 0
        self.h, self.dh, self.d = num_heads, d_model // num_heads, d_model
        self.W_q = nn.Linear(d_model, d_model)
        self.W_k = nn.Linear(d_model, d_model)
        self.W_v = nn.Linear(d_model, d_model)
        self.W_o = nn.Linear(d_model, d_model)

    def split(self, x):
        B, L, _ = x.shape
        return x.view(B, L, self.h, self.dh).transpose(1, 2)

    def forward(self, x_q, x_kv, mask=None):
        Q, K, V = self.split(self.W_q(x_q)), self.split(self.W_k(x_kv)), self.split(self.W_v(x_kv))
        scores = Q @ K.transpose(-2, -1) / math.sqrt(self.dh)
        if mask is not None:
            scores = scores.masked_fill(~mask, float("-inf"))
        w = torch.softmax(scores, dim=-1)
        out = (w @ V).transpose(1, 2).contiguous().view(x_q.size(0), x_q.size(1), self.d)
        return self.W_o(out), w

class AddNorm(nn.Module):
    def __init__(self, d_model, dropout=0.1):
        super().__init__()
        self.norm = nn.LayerNorm(d_model)
        self.drop = nn.Dropout(dropout)
    def forward(self, x, sub):
        return self.norm(x + self.drop(sub))

class FeedForward(nn.Module):
    def __init__(self, d_model, d_ff, dropout=0.1):
        super().__init__()
        self.net = nn.Sequential(nn.Linear(d_model, d_ff), nn.GELU(),
                                 nn.Dropout(dropout), nn.Linear(d_ff, d_model))
    def forward(self, x):
        return self.net(x)

class EncoderBlock(nn.Module):
    def __init__(self, d_model, heads, d_ff, dropout):
        super().__init__()
        self.attn, self.an1 = MultiHeadAttention(d_model, heads), AddNorm(d_model, dropout)
        self.ffn,  self.an2 = FeedForward(d_model, d_ff, dropout), AddNorm(d_model, dropout)
    def forward(self, x, src_mask):
        x = self.an1(x, self.attn(x, x, src_mask)[0])
        return self.an2(x, self.ffn(x))

class DecoderBlock(nn.Module):
    def __init__(self, d_model, heads, d_ff, dropout):
        super().__init__()
        self.self_attn,  self.an1 = MultiHeadAttention(d_model, heads), AddNorm(d_model, dropout)
        self.cross_attn, self.an2 = MultiHeadAttention(d_model, heads), AddNorm(d_model, dropout)
        self.ffn,        self.an3 = FeedForward(d_model, d_ff, dropout), AddNorm(d_model, dropout)
    def forward(self, x, enc, tgt_mask, src_mask):
        x = self.an1(x, self.self_attn(x, x, tgt_mask)[0])
        x = self.an2(x, self.cross_attn(x, enc, src_mask)[0])
        return self.an3(x, self.ffn(x))

class Transformer(nn.Module):
    def __init__(self, vocab, d_model=384, heads=12, d_ff=1536, layers=3, max_len=MAX_LEN, dropout=0.1):
        super().__init__()
        self.d_model = d_model
        self.emb = nn.Embedding(vocab, d_model, padding_idx=PAD_ID)   # shared for en + bn
        self.pos = nn.Embedding(max_len, d_model)
        self.drop = nn.Dropout(dropout)
        self.enc_layers = nn.ModuleList([EncoderBlock(d_model, heads, d_ff, dropout) for _ in range(layers)])
        self.dec_layers = nn.ModuleList([DecoderBlock(d_model, heads, d_ff, dropout) for _ in range(layers)])
        self.out = nn.Linear(d_model, vocab)

    def embed(self, ids):
        pos = torch.arange(ids.size(1), device=ids.device).unsqueeze(0)
        return self.drop(self.emb(ids) * math.sqrt(self.d_model) + self.pos(pos))

    def encode(self, src, src_mask):
        x = self.embed(src)
        for l in self.enc_layers:
            x = l(x, src_mask)
        return x

    def decode(self, tgt_in, enc, tgt_mask, src_mask):
        x = self.embed(tgt_in)
        for l in self.dec_layers:
            x = l(x, enc, tgt_mask, src_mask)
        return self.out(x)

    def forward(self, src, tgt_in):
        src_mask = (src != PAD_ID)[:, None, None, :]                      
        T = tgt_in.size(1)
        causal = torch.tril(torch.ones(T, T, device=src.device, dtype=torch.bool))
        tgt_mask = (tgt_in != PAD_ID)[:, None, None, :] & causal           
        enc = self.encode(src, src_mask)
        return self.decode(tgt_in, enc, tgt_mask, src_mask)               
```

```python
#xavier initialization 
def init_weights(model):
    for name, p in model.named_parameters():
        if p.dim() > 1:
            nn.init.xavier_uniform_(p)
        elif "bias" in name:
            nn.init.zeros_(p)
    with torch.no_grad():
        model.emb.weight[PAD_ID].zero_()

def count_correct(logits, target):
    """Returns (correct tokens, total non-pad tokens) for one batch."""
    pred = logits.argmax(dim=-1)
    mask = target != PAD_ID
    return ((pred == target) & mask).sum().item(), mask.sum().item()

DROPOUT = 0.5

model = Transformer(VOCAB_SIZE, dropout=DROPOUT).to(DEVICE)
init_weights(model)
optimizer = torch.optim.AdamW(model.parameters(), lr=1e-4, weight_decay=0.09)
criterion = nn.CrossEntropyLoss(ignore_index=PAD_ID, label_smoothing=0.5)
EPOCHS = 60

for epoch in range(EPOCHS):
    model.train()
    total, tr_correct, tr_tokens = 0.0, 0, 0
    for src, tgt_in, tgt_out in train_loader:
        src, tgt_in, tgt_out = src.to(DEVICE), tgt_in.to(DEVICE), tgt_out.to(DEVICE)
        logits = model(src, tgt_in)
        loss = criterion(logits.view(-1, VOCAB_SIZE), tgt_out.view(-1))
        optimizer.zero_grad()
        loss.backward()
        torch.nn.utils.clip_grad_norm_(model.parameters(), 1.0)
        optimizer.step()
        total += loss.item()
        c, n = count_correct(logits.detach(), tgt_out)
        tr_correct += c
        tr_tokens += n

    model.eval()
    val, va_correct, va_tokens = 0.0, 0, 0
    with torch.no_grad():
        for src, tgt_in, tgt_out in val_loader:
            src, tgt_in, tgt_out = src.to(DEVICE), tgt_in.to(DEVICE), tgt_out.to(DEVICE)
            logits = model(src, tgt_in)
            val += criterion(logits.view(-1, VOCAB_SIZE), tgt_out.view(-1)).item()
            c, n = count_correct(logits, tgt_out)
            va_correct += c
            va_tokens += n

    train_acc = 100 * tr_correct / tr_tokens
    val_acc = 100 * va_correct / va_tokens
    print(f"Epoch {epoch+1}/{EPOCHS}  "
          f"train loss {total/len(train_loader):.4f} acc {train_acc:.2f}%  |  "
          f"val loss {val/len(val_loader):.4f} acc {val_acc:.2f}%")

torch.save(model.state_dict(), "en_bn_transformer.pt")
```

```python
@torch.no_grad()
def token_accuracy(loader):
    model.eval()
    correct = total = 0
    for src, tgt_in, tgt_out in loader:
        src, tgt_in, tgt_out = src.to(DEVICE), tgt_in.to(DEVICE), tgt_out.to(DEVICE)
        pred = model(src, tgt_in).argmax(-1)
        mask = tgt_out != PAD_ID
        correct += ((pred == tgt_out) & mask).sum().item()
        total += mask.sum().item()
    return correct / total

print("val token accuracy:", token_accuracy(val_loader))
```

```python
@torch.no_grad()
def translate(sentence, max_new=MAX_LEN):
    model.eval()
    src = torch.tensor([sp.encode(sentence)[:MAX_LEN - 1] + [EOS_ID]], device=DEVICE)
    src_mask = (src != PAD_ID)[:, None, None, :]
    enc = model.encode(src, src_mask)
    ys = torch.tensor([[BOS_ID]], device=DEVICE)
    for _ in range(max_new):
        T = ys.size(1)
        tgt_mask = torch.tril(torch.ones(T, T, device=DEVICE, dtype=torch.bool))[None, None]
        next_id = model.decode(ys, enc, tgt_mask, src_mask)[:, -1].argmax(-1, keepdim=True)
        ys = torch.cat([ys, next_id], dim=1)
        if next_id.item() == EOS_ID:
            break
    return sp.decode(ys[0, 1:].tolist())

print(translate("i am an engineer ."))
```
